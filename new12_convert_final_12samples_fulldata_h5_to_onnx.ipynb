# Convert Final 12-Sample Fulldata H5 To ONNX

This notebook converts the final deployment H5 model for `fulldata_final_cnn_only` into ONNX.


```python
from pathlib import Path
import json

PROJECT_ROOT = Path.cwd()
MODELS_ROOT = PROJECT_ROOT / "training_outputs" / "mobilenetv3small_12fold_fulldata_final_cnn_only" / "models"

MODEL_H5_PATH = MODELS_ROOT / "meatlens_best_fulldata_final_cnn_only_mobilenetv3small.h5"
MODEL_ONNX_PATH = MODELS_ROOT / "meatlens_best_fulldata_final_cnn_only_mobilenetv3small.onnx"

print("MODEL_H5_PATH =", MODEL_H5_PATH)
print("MODEL_ONNX_PATH =", MODEL_ONNX_PATH)
print("H5 exists?", MODEL_H5_PATH.exists())
```

```python
import tensorflow as tf
import tf2onnx

def convert_h5_to_onnx():
    model = tf.keras.models.load_model(MODEL_H5_PATH, compile=False)
    input_signature = (
        tf.TensorSpec((None, 224, 224, 3), tf.float32, name="image_input"),
    )
    model_proto, external_tensor_storage = tf2onnx.convert.from_keras(
        model,
        input_signature=input_signature,
        opset=13,
        output_path=str(MODEL_ONNX_PATH),
    )
    print("Successfully converted to ONNX!")
    print("Size (MB):", MODEL_ONNX_PATH.stat().st_size / (1024 * 1024))

convert_h5_to_onnx()
```
